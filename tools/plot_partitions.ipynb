```
import matplotlib.pyplot as plt
from matplotlib import colormaps
import pandas as pd

partitions = pd.read_csv("../build/partitions.csv", sep=', ', engine='python')


for l in partitions['level'].unique():
  particles = partitions[partitions['level'] == l]
  plt.figure()
  plt.xlim(-110, 110),
  plt.ylim(-110, 110)
  c_map = colormaps['jet']
  for p_id in particles['partition_id'].unique():
    p_id_tmp = p_id
    color = 0
    power = 0.25
    while p_id_tmp != 0:
      color = color + (p_id_tmp % 4) * power
      p_id_tmp = p_id_tmp // 4
      power = power / 4
    partition = particles[particles['partition_id'] == p_id]
    plt.scatter(partition['x'], partition['y'], color=c_map(color), label=f'Partition {p_id}')
    plt.title(f'Node level {l}' + (' / root' if l == 0 else ''))
    plt.legend(loc='lower right')
  plt.show()
```
