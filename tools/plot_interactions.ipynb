```
import matplotlib.pyplot as plt
import pandas as pd

interactions = pd.read_csv("../build/interactions.csv", sep=', ', engine='python')

# Define your color palette
color_map = {
    'x': 'tab:red',
    'n': 'tab:blue',
    'i': 'tab:green'
}

# 1. Pre-filter the baseline subset (target=0, depth=0)
baseline = interactions[(interactions['target'] == 0) & (interactions['depth'] == 0)]

# Group by target and depth together
for (target_val, depth_val), particles in interactions.groupby(['target', 'depth']):

    
    plt.figure()

    # Find baseline points that are NOT in the current group
    # (Uses index exclusion so matching row IDs are dropped)
    other_baseline = baseline[~baseline.index.isin(particles.index)]

    # Plot baseline reference points in the background
    if not other_baseline.empty:
        plt.scatter(
            other_baseline['x'],
            other_baseline['y'],
            c='lightgray',
            s=20,
            label='Target 0, Depth 0 (other)',
            zorder=1
        )

    plt.xlim(-110, 110)
    plt.ylim(-110, 110)
    plt.scatter(
        particles['x'], 
        particles['y'], 
        c=particles['type'].map(color_map)
    )
    plt.title(f"Target: {target_val}, Depth: {depth_val}")
    plt.xlabel('x')
    plt.ylabel('y')
    plt.show()
```
